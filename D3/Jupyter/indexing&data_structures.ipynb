## 1-**Tuple**
- ordered collection of elements
- enclosed in () round braces/parenthesis
- different kind of elements can be stored
- Once Elements are stored you can't change them (unmutatable)

```python
tup1 = (1, "Python", True, 2.5)
tup1
```

```python
# type of a tuple
type(tup1)
```

### - Indexing in tuple

```python
tup1[1]
```

```python
tup1[2]
```

```python
# last element is exclusive
tup1[0:3]
```

```python
# count of elements in tuple
len(tup1)
```

```python
tup2 = (2, "tauhaazmat", 3.5, False)
tup2
```

```python
# concatinate (to add two or more than two tuples)
tup1 + tup2
```

```python
# concatinate + repeat
tup1*2 + tup2
```

```python
tup3 = (20, 30, 50, 60, 80, 100)
tup3
```

```python
# minimum
min(tup3)
```

```python
# maximum
max(tup3)
```

```python
tup3 * 2
```

## 3- Dictionaries
- An unordered collection of elements
- Key and Value
- curly braces or brackets {}
- mutatable/change the values

```python
# Food and their prices
food1 = {"Pasta":30, "Burger":100, "Pizza":20, "ColeSlaw":50, "Chicken Wings":50}
food1
```

```python
type(food1)
```

```python
# extract data
keys1 = food1.keys()
keys1
```

```python
values = food1.values()
values
```

```python
# Adding new element
food1["Patties"]=10
food1
```

```python
# update the values
food1["Patties"]=15
food1
```

```python
food2 = {"Dates":50, "Chocolates":200, "Alfredo Pasta":1000}
food2
```

```python
# Concatinate
food1.update(food2)
food1
```
