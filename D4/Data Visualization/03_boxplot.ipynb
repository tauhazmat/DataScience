```python
# import seaborn library
import seaborn
# canvas (Baloon board)
seaborn.set(style='whitegrid')
# load titanic dataset
ship = seaborn.load_dataset("titanic")

seaborn.boxplot(x="class", 
               y="fare",
               data=ship)
```
